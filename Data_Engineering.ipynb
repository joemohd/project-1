```python
import pandas as pd
import glob
import os

# Locate the two CSV files
files = glob.glob('/content/*.csv')

housing_file = next(f for f in files if 'Affordable_Housing' in os.path.basename(f))
transport_file = next(f for f in files if 'Serious_Injury' in os.path.basename(f))

# Load datasets
df_housing = pd.read_csv(housing_file, low_memory=False)
df_transport = pd.read_csv(transport_file, low_memory=False)

print("HOUSING DATASET")
print("Shape:", df_housing.shape)
display(df_housing.head())

print("\nTRANSPORTATION DATASET")
print("Shape:", df_transport.shape)
display(df_transport.head())
```

```python
# Dataset dimensions
print("HOUSING DATASET")
print("Rows and columns:", df_housing.shape)
print("\nColumn names:")
print(df_housing.columns.tolist())

print("\n" + "="*60)

print("TRANSPORTATION DATASET")
print("Rows and columns:", df_transport.shape)
print("\nColumn names:")
print(df_transport.columns.tolist())
```

```python
def explore_dataset(df, name):

    print(f"\n{'='*60}")
    print(f"EXPLORING: {name}")
    print(f"{'='*60}")

    # Dataset dimensions
    print("\n1. Dataset Shape:")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")

    # Column names and data types
    print("\n2. Column Information:")
    display(pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values,
        "Missing Values": df.isnull().sum().values,
        "Missing %": (df.isnull().mean().values * 100).round(2)
    }))

    # Duplicate records
    print("\n3. Duplicate Records:")
    print(df.duplicated().sum())

    # Descriptive statistics
    print("\n4. Descriptive Statistics:")
    display(df.describe(include="all").T)

    # Missing values
    print("\n5. Missing Values:")
    missing = df.isnull().sum()
    display(missing[missing > 0].sort_values(ascending=False))


explore_dataset(df_housing, "HOUSING DATASET")
explore_dataset(df_transport, "TRANSPORTATION DATASET")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Compare missing values in both datasets
for df, name in [
    (df_housing, "Housing Dataset"),
    (df_transport, "Transportation Dataset")
]:

    missing = df.isnull().mean().mul(100)
    missing = missing[missing > 0].sort_values(ascending=False)

    plt.figure(figsize=(12, 5))

    if len(missing) > 0:
        sns.barplot(x=missing.index, y=missing.values)
        plt.xticks(rotation=90)
        plt.ylabel("Missing Values (%)")
        plt.title(f"Missing Values - {name}")
    else:
        plt.text(0.5, 0.5, "No Missing Values",
                 ha="center", va="center")
        plt.title(name)

    plt.tight_layout()
    plt.show()
```

```python
for df, name in [
    (df_housing, "Housing"),
    (df_transport, "Transportation")
]:

    numeric_columns = df.select_dtypes(include="number").columns

    print(f"\n{name} Numerical Distributions")

    df[numeric_columns].hist(
        figsize=(14, 10),
        bins=30
    )

    plt.tight_layout()
    plt.show()
```

```python
# Display all columns with their data types

for df, name in [
    (df_housing, "HOUSING"),
    (df_transport, "TRANSPORTATION")
]:
    print(f"\n{'='*50}")
    print(name)
    print('='*50)

    for col in df.columns:
        print(f"{col:40} {df[col].dtype}")
```

```python
# Search for geographic and date-related columns

keywords = [
    'borough', 'zip', 'address', 'latitude',
    'longitude', 'date', 'year', 'month',
    'location', 'community'
]

for df, name in [
    (df_housing, "HOUSING"),
    (df_transport, "TRANSPORTATION")
]:
    print(f"\n{name} POTENTIAL INTEGRATION COLUMNS")

    for col in df.columns:
        if any(word in col.lower() for word in keywords):
            print(f"\nColumn: {col}")
            display(df[col].head(5))
```

```python
for df, name in [
    (df_housing, "HOUSING"),
    (df_transport, "TRANSPORTATION")
]:
    print(f"\n{name} OUTLIER ANALYSIS")

    numeric = df.select_dtypes(include='number')

    for col in numeric.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = ((df[col] < lower) | (df[col] > upper)).sum()

        if outliers > 0:
            print(f"{col}: {outliers:,} potential outliers")
```

```python
# Preserve original datasets
housing_clean = df_housing.copy()
transport_clean = df_transport.copy()

print("Working copies created successfully!")

print("Housing:", housing_clean.shape)
print("Transportation:", transport_clean.shape)
```

```python
print("HOUSING DUPLICATES:")
print(housing_clean.duplicated().sum())

print("\nTRANSPORTATION DUPLICATES:")
print(transport_clean.duplicated().sum())
```

```python
housing_clean = housing_clean.drop_duplicates()
transport_clean = transport_clean.drop_duplicates()

print("Housing after cleaning:", housing_clean.shape)
print("Transportation after cleaning:", transport_clean.shape)
```

```python
print("HOUSING:")
display(housing_clean.isnull().sum().sort_values(ascending=False).head(15))

print("\nTRANSPORTATION:")
display(transport_clean.isnull().sum().sort_values(ascending=False).head(15))
```

```python
def missing_analysis(df, name):
    print(f"\n===== {name} =====")

    report = pd.DataFrame({
        "Missing": df.isnull().sum(),
        "Percentage": (df.isnull().mean() * 100).round(2),
        "Data Type": df.dtypes.astype(str)
    })

    report = report[report["Missing"] > 0]
    display(report.sort_values("Percentage", ascending=False))


missing_analysis(housing_clean, "HOUSING")
missing_analysis(transport_clean, "TRANSPORTATION")
```

```python
# Standardize column names
for df in [housing_clean, transport_clean]:
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(r'[^a-z0-9]+', '_', regex=True)
        .str.strip('_')
    )

print("HOUSING COLUMNS:")
print(housing_clean.columns.tolist())

print("\nTRANSPORTATION COLUMNS:")
print(transport_clean.columns.tolist())
```

```python
for df, name in [
    (housing_clean, "HOUSING"),
    (transport_clean, "TRANSPORTATION")
]:
    print(f"\n{name} DATE COLUMNS:")

    for col in df.columns:
        if 'date' in col or 'time' in col:
            print(col, "|", df[col].dtype)
            print(df[col].head(3).tolist())
```

```python
# Convert columns containing date or time information

for df, name in [
    (housing_clean, "Housing"),
    (transport_clean, "Transportation")
]:
    print(f"\n{name}")

    for col in df.columns:
        if 'date' in col or 'time' in col:
            df[col] = pd.to_datetime(df[col], errors='coerce')

            print(
                col,
                "| Converted to:",
                df[col].dtype,
                "| Invalid dates:",
                df[col].isna().sum()
            )
```

```python
# Check borough values
for df, name in [
    (housing_clean, "HOUSING"),
    (transport_clean, "TRANSPORTATION")
]:
    print(f"\n===== {name} =====")

    for col in df.columns:
        if 'borough' in col:
            print(f"\n{col}")
            print(df[col].value_counts(dropna=False).head(20))

        if 'zip' in col:
            print(f"\n{col}")
            print("Unique ZIP codes:", df[col].nunique())
            print(df[col].head(10).tolist())
```

```python
# Standardize borough names
for df in [housing_clean, transport_clean]:

    for col in df.columns:
        if 'borough' in col:
            df[col] = (
                df[col]
                .astype('string')
                .str.strip()
                .str.upper()
            )

            df[col] = df[col].replace({
                'MANHATTAN': 'MANHATTAN',
                'BROOKLYN': 'BROOKLYN',
                'QUEENS': 'QUEENS',
                'BRONX': 'BRONX',
                'STATEN ISLAND': 'STATEN ISLAND'
            })

print("Borough standardization completed.")
```

```python
for df, name in [
    (housing_clean, "HOUSING"),
    (transport_clean, "TRANSPORTATION")
]:
    print(f"\n{name}")

    for col in df.columns:
        if 'borough' in col:
            print(df[col].value_counts(dropna=False))
```

```python
# Generate a final data quality report

for df, name in [
    (housing_clean, "HOUSING"),
    (transport_clean, "TRANSPORTATION")
]:
    print(f"\n===== {name} =====")

    report = pd.DataFrame({
        "Missing": df.isnull().sum(),
        "Missing %": (df.isnull().mean() * 100).round(2),
        "Unique Values": df.nunique()
    })

    display(report.sort_values("Missing %", ascending=False))
```

```python
for df, name in [
    (housing_clean, "HOUSING"),
    (transport_clean, "TRANSPORTATION")
]:
    print(f"\n===== {name} =====")

    for col in df.select_dtypes(include='number').columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        count = ((df[col] < lower) | (df[col] > upper)).sum()

        print(f"{col}: {count:,} potential outliers")
```

```python
for df, name in [
    (housing_clean, "HOUSING"),
    (transport_clean, "TRANSPORTATION")
]:
    print(f"\n{'='*50}")
    print(name)
    print('='*50)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Duplicates:", df.duplicated().sum())
    print("Missing cells:", df.isnull().sum().sum())
    print("Memory usage (MB):",
          round(df.memory_usage(deep=True).sum() / 1024**2, 2))

    print("\nMissing values by column:")
    display(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Final validation before export

for df, name in [
    (housing_clean, "Housing"),
    (transport_clean, "Transportation")
]:
    print(f"\n{name}")
    print("Shape:", df.shape)
    print("Duplicate rows:", df.duplicated().sum())
    print("Missing cells:", df.isnull().sum().sum())
    print("Column names unique:", df.columns.is_unique)
```

```python
housing_clean.to_csv(
    "Housing_Cleaned.csv", index=False
)

transport_clean.to_csv(
    "Transportation_Cleaned.csv", index=False
)

print("Export completed!")
```
